```python
import requests

def obtener_ruta(origen, destino):

    lon1 = origen[1]
    lat1 = origen[0]

    lon2 = destino[1]
    lat2 = destino[0]

    url = (
        f"https://router.project-osrm.org/"
        f"route/v1/driving/"
        f"{lon1},{lat1};{lon2},{lat2}"
        f"?overview=false"
    )

    respuesta = requests.get(url)

    datos_ruta = respuesta.json()

    if datos_ruta["code"] == "Ok":

        ruta = datos_ruta["routes"][0]

        distancia = ruta["distance"] / 1000
        tiempo = ruta["duration"] / 60

        return distancia, tiempo

    else:

        return None, None
```

### Solución rápida y eficiente: OSRM Table API
La API `/table` de OSRM nos permite calcular la matriz de tiempo de viaje de todos contra todos en una sola llamada de red en lugar de hacer 992 llamadas individuales.

```python
import requests
import numpy as np

def obtener_matrices_osrm(df):
    # Formatear todas las coordenadas en el formato: lon,lat;lon,lat;...
    coordenadas = ";".join([f"{lon},{lat}" for lat, lon in zip(df['Latitud'], df['Longitud'])])

    # Usamos el servicio 'table' para obtener distancias y tiempos de todos contra todos
    url = f"https://router.project-osrm.org/table/v1/driving/{coordenadas}?annotations=duration,distance"

    respuesta = requests.get(url)
    datos_tabla = respuesta.json()

    if datos_tabla.get("code") == "Ok":
        # OSRM devuelve distancias en metros y duraciones en segundos
        matriz_distancia = np.array(datos_tabla["distances"]) / 1000.0  # Convertir a km
        matriz_tiempo = np.array(datos_tabla["durations"]) / 60.0       # Convertir a minutos
        return matriz_distancia, matriz_tiempo
    else:
        print("Error en la API de OSRM:", datos_tabla.get("code"))
        return None, None

# Ejecutar la consulta masiva (se resolverá en un par de segundos)
matriz_distancia_rapida, matriz_tiempo_rapida = obtener_matrices_osrm(df)

if matriz_distancia_rapida is not None:
    print("¡Matrices generadas instantáneamente!")
    print("Formato de la matriz:", matriz_distancia_rapida.shape)
```

### Optimización de la Ruta (TSP) con Google OR-Tools
Definimos el centro de distribución en **León** (índice `0`) y resolvemos para encontrar la ruta que minimiza el tiempo total de viaje pasando por todos los puntos.

```python
from ortools.constraint_solver import routing_enums_pb2
from ortools.constraint_solver import pywrapcp

def resolver_tsp(matriz_tiempo, de_index=0):
    # Inicializar el gestor de índices de enrutamiento: 32 locaciones, 1 vehículo, salida y llegada en 'de_index'
    manager = pywrapcp.RoutingIndexManager(len(matriz_tiempo), 1, de_index)
    routing = pywrapcp.RoutingModel(manager)

    # Crear la función de tránsito (costo basado en el tiempo)
    def callback_tiempo(from_index, to_index):
        from_node = manager.IndexToNode(from_index)
        to_node = manager.IndexToNode(to_index)
        return int(matriz_tiempo[from_node][to_node] * 100)  # Multiplicamos para trabajar con enteros

    transit_callback_index = routing.RegisterTransitCallback(callback_tiempo)
    routing.SetArcCostEvaluatorOfAllVehicles(transit_callback_index)

    # Configurar los parámetros de búsqueda
    search_parameters = pywrapcp.DefaultRoutingSearchParameters()
    search_parameters.first_solution_strategy = (
        routing_enums_pb2.FirstSolutionStrategy.PATH_CHEAPEST_ARC
    )

    # Resolver el problema
    solucion = routing.SolveWithParameters(search_parameters)

    if solucion:
        # Extraer el orden de la ruta
        index = routing.Start(0)
        ruta_indices = []
        while not routing.IsEnd(index):
            ruta_indices.append(manager.IndexToNode(index))
            index = solucion.Value(routing.NextVar(index))
        ruta_indices.append(manager.IndexToNode(index))
        return ruta_indices
    else:
        return None

# Resolver la ruta optimizada usando la matriz de tiempos de OSRM
indice_leon = 0
orden_ruta = resolver_tsp(matriz_tiempo_rapida, de_index=indice_leon)

# Mostrar el itinerario óptimo
if orden_ruta:
    print("--- ITINERARIO OPTIMIZADO DE VIAJE ---\n")
    tiempo_acumulado = 0
    distancia_acumulada = 0

    for paso, idx in enumerate(orden_ruta):
        ciudad = df.iloc[idx]['Ciudad']
        estado = df.iloc[idx]['Estado']

        if paso == 0:
            print(f"Inicio: {ciudad} ({estado})")
        else:
            idx_anterior = orden_ruta[paso - 1]
            t_tramo = matriz_tiempo_rapida[idx_anterior][idx]
            d_tramo = matriz_distancia_rapida[idx_anterior][idx]
            tiempo_acumulado += t_tramo
            distancia_acumulada += d_tramo
            print(f"  -> {ciudad} ({estado}) [Tramo: {d_tramo:.1f} km, {t_tramo/60:.1f} hrs]")

    print(f"\nResumen:")
    print(f"- Distancia total: {distancia_acumulada:.2f} km")
    print(f"- Tiempo total de conducción: {tiempo_acumulado/60:.2f} horas")
else:
    print("No se pudo encontrar una ruta viable.")
```

### Trazado de Ruta Real por Carretera
Consultamos la geometría de carretera real a OSRM para la secuencia ordenada del itinerario y la sobreponemos en el mapa.

```python
import requests
import folium
from folium.plugins import AntPath

def obtener_geometria_ruta_completa(orden_ruta, df):
    # Obtener las coordenadas en el orden optimizado
    coords_ordenadas = [f"{df.iloc[idx]['Longitud']},{df.iloc[idx]['Latitud']}" for idx in orden_ruta]

    # Debido a límites de longitud de URL, podemos solicitar el trazado por tramos
    # o intentar todo junto si el servidor lo soporta (32 puntos suele ser aceptado por OSRM)
    coordenadas_str = ";".join(coords_ordenadas)

    url = f"https://router.project-osrm.org/route/v1/driving/{coordenadas_str}?overview=full&geometries=geojson"

    respuesta = requests.get(url)
    datos_ruta = respuesta.json()

    if datos_ruta.get("code") == "Ok":
        # Extraer la geometría en formato GeoJSON (coordenadas como [lon, lat])
        geom = datos_ruta["routes"][0]["geometry"]["coordinates"]
        # Invertir a [lat, lon] para Folium
        ruta_carretera = [[lat, lon] for lon, lat in geom]
        return ruta_carretera
    else:
        print("Error al obtener la geometría detallada. Usando trazado punto a punto por tramos...")
        # Alternativa paso a paso si la ruta completa falla por tamaño
        ruta_carretera = []
        for i in range(len(orden_ruta) - 1):
            orig = coords_ordenadas[i]
            dest = coords_ordenadas[i+1]
            url_tramo = f"https://router.project-osrm.org/route/v1/driving/{orig};{dest}?overview=full&geometries=geojson"
            resp_tramo = requests.get(url_tramo).json()
            if resp_tramo.get("code") == "Ok":
                geom = resp_tramo["routes"][0]["geometry"]["coordinates"]
                ruta_carretera.extend([[lat, lon] for lon, lat in geom])
        return ruta_carretera

# Obtener los puntos de carretera reales
puntos_carretera = obtener_geometria_ruta_completa(orden_ruta, df)

# Crear el mapa interactivo
lat_centro = df['Latitud'].mean()
lon_centro = df['Longitud'].mean()
mapa_carretera = folium.Map(location=[lat_centro, lon_centro], zoom_start=6)

# Dibujar la ruta animada siguiendo las curvas de la carretera
if puntos_carretera:
    AntPath(
        locations=puntos_carretera,
        dash_array=[10, 20],
        delay=1000,
        color="#0078FF",
        pulse_color="#FFD700",
        weight=5,
        opacity=0.85,
        tooltip="Ruta Real por Carretera"
    ).add_to(mapa_carretera)

# Agregar los marcadores de las paradas
for paso, idx in enumerate(orden_ruta[:-1]):
    fila = df.iloc[idx]
    if idx == indice_leon:
        color_icono = "red"
        prefijo = "Inicio/Fin"
        icono = "play"
    else:
        color_icono = "blue"
        prefijo = f"Parada {paso}"
        icono = "info-sign"

    folium.Marker(
        location=[fila['Latitud'], fila['Longitud']],
        popup=f"<b>{prefijo}</b><br>{fila['Ciudad']}<br>{fila['Estado']}",
        tooltip=f"{paso}. {fila['Ciudad']}",
        icon=folium.Icon(color=color_icono, icon=icono)
    ).add_to(mapa_carretera)

mapa_carretera
```

### Agrupación y Optimización por Zonas Regionales
Definimos los 3 grupos solicitados, asignamos las ciudades a su zona correspondiente, calculamos el orden de ruta óptimo para cada una y trazamos sus trayectos reales por carretera.

```python
import requests
import numpy as np
import folium
from folium.plugins import AntPath
from ortools.constraint_solver import routing_enums_pb2
from ortools.constraint_solver import pywrapcp

# --- 1. Definir la lógica de asignación a cada una de las 4 nuevas zonas ---
grupo_sur_oriente = []    # Colima, Michoacán, Oaxaca
grupo_centro_sur = []     # Estado de México, Hidalgo, Puebla, Morelos
grupo_sur_occidente = []  # Veracruz, Tabasco, Chiapas
grupo_yucatan = []          # Yucatán, Quintana Roo

for idx, fila in df.iterrows():
    if idx == 0:
        continue

    estado = fila['Estado']

    if estado in ['Colima', 'Michoacán', 'Oaxaca']:
        grupo_sur_oriente.append(idx)
    elif estado in ['Estado de México', 'Hidalgo', 'Puebla', 'Morelos', 'Ciudad de México']:
        # Incluimos Ciudad de México en Centro Sur por cercanía geográfica natural rodeada por Edomex/Hidalgo/Morelos
        grupo_centro_sur.append(idx)
    elif estado in ['Veracruz', 'Tabasco', 'Chiapas']:
        grupo_sur_occidente.append(idx)
    elif estado in ['Yucatán', 'Quintana Roo']:
        grupo_yucatan.append(idx)

zonas = {
    "Sur Oriente": grupo_sur_oriente,
    "Centro Sur": grupo_centro_sur,
    "Sur Occidente": grupo_sur_occidente,
    "Península de Yucatán": grupo_yucatan
}

# --- 2. Función genérica para resolver el TSP de un grupo específico ---
def optimizar_zona(indices_grupo, matriz_tiempo):
    indices_ruta = [0] + indices_grupo
    n_local = len(indices_ruta)

    manager = pywrapcp.RoutingIndexManager(n_local, 1, 0)
    routing = pywrapcp.RoutingModel(manager)

    def callback_tiempo(from_index, to_index):
        from_node = indices_ruta[manager.IndexToNode(from_index)]
        to_node = indices_ruta[manager.IndexToNode(to_index)]
        return int(matriz_tiempo[from_node][to_node] * 100)

    transit_callback_index = routing.RegisterTransitCallback(callback_tiempo)
    routing.SetArcCostEvaluatorOfAllVehicles(transit_callback_index)

    search_parameters = pywrapcp.DefaultRoutingSearchParameters()
    search_parameters.first_solution_strategy = (
        routing_enums_pb2.FirstSolutionStrategy.PATH_CHEAPEST_ARC
    )

    solucion = routing.SolveWithParameters(search_parameters)

    if solucion:
        index = routing.Start(0)
        ruta_ordenada = []
        while not routing.IsEnd(index):
            ruta_ordenada.append(indices_ruta[manager.IndexToNode(index)])
            index = solucion.Value(routing.NextVar(index))
        ruta_ordenada.append(indices_ruta[manager.IndexToNode(index)])
        return ruta_ordenada
    return indices_ruta + [0]

# --- 3. Calcular las rutas óptimas e imprimir itinerario ---
rutas_regionales = {}
for nombre_zona, indices in zonas.items():
    rutas_regionales[nombre_zona] = optimizar_zona(indices, matriz_tiempo_rapida)

# --- 4. Trazar en un mapa interactivo con carreteras reales ---
colores_zonas = {
    "Sur Oriente": "#75FF33",       # Verde lima
    "Centro Sur": "#33C1FF",        # Celeste
    "Sur Occidente": "#FF33F6",     # Rosa/Fucsia
    "Península de Yucatán": "#FFC300" # Amarillo dorado
}

mapa_zonas = folium.Map(location=[df['Latitud'].mean(), df['Longitud'].mean()], zoom_start=6)

print("--- PLAN DE DISTRIBUCIÓN POR REGIONES OPTIMIZADAS ---\n")
for nombre_zona, ruta in rutas_regionales.items():
    color = colores_zonas[nombre_zona]

    # Solicitar geometría de OSRM para esta sub-ruta
    coords = [f"{df.iloc[idx]['Longitud']},{df.iloc[idx]['Latitud']}" for idx in ruta]
    url = f"https://router.project-osrm.org/route/v1/driving/{';'.join(coords)}?overview=full&geometries=geojson"
    resp = requests.get(url).json()

    puntos_carretera = []
    if resp.get("code") == "Ok":
        geom = resp["routes"][0]["geometry"]["coordinates"]
        puntos_carretera = [[lat, lon] for lon, lat in geom]

    # Dibujar la carretera en el mapa
    if puntos_carretera:
        AntPath(
            locations=puntos_carretera,
            delay=1000,
            color=color,
            pulse_color="#FFFFFF",
            weight=5,
            opacity=0.85,
            tooltip=f"Zona: {nombre_zona}"
        ).add_to(mapa_zonas)

    # Imprimir en texto
    itinerario = [df.iloc[i]['Ciudad'] for i in ruta]
    print(f"🔹 ZONA: {nombre_zona} ({len(ruta)-2} paradas)")
    print(" -> ".join(itinerario))
    print("-" * 50)

# Añadir los marcadores en el mapa
folium.Marker(
    location=[df.iloc[0]['Latitud'], df.iloc[0]['Longitud']],
    popup="<b>Centro de Distribución (León)</b>",
    icon=folium.Icon(color="red", icon="briefcase")
).add_to(mapa_zonas)

for nombre_zona, ruta in rutas_regionales.items():
    color_marcador = "green" if nombre_zona == "Sur Oriente" else "blue" if nombre_zona == "Centro Sur" else "purple" if nombre_zona == "Sur Occidente" else "orange"
    for paso, idx in enumerate(ruta[1:-1], 1):
        fila = df.iloc[idx]
        folium.Marker(
            location=[fila['Latitud'], fila['Longitud']],
            popup=f"<b>{nombre_zona}</b><br>{fila['Ciudad']}<br>Parada {paso}",
            tooltip=f"{nombre_zona}: {fila['Ciudad']}",
            icon=folium.Icon(color=color_marcador, icon="info-sign")
        ).add_to(mapa_zonas)

mapa_zonas
```

```python
import numpy as np
import folium
from folium.plugins import AntPath
import requests
from ortools.constraint_solver import routing_enums_pb2
from ortools.constraint_solver import pywrapcp

def resolver_vrp_ultra_eficiente(num_vehiculos):
    # Inicializamos el gestor de enrutamiento con los datos de tiempo de viaje de OSRM
    manager = pywrapcp.RoutingIndexManager(len(matriz_tiempo_rapida), num_vehiculos, 0)
    routing = pywrapcp.RoutingModel(manager)

    def callback_tiempo(from_index, to_index):
        from_node = manager.IndexToNode(from_index)
        to_node = manager.IndexToNode(to_index)
        return int(matriz_tiempo_rapida[from_node][to_node] * 100)

    transit_callback_index = routing.RegisterTransitCallback(callback_tiempo)
    routing.SetArcCostEvaluatorOfAllVehicles(transit_callback_index)

    # Agregamos una dimensión de tiempo sin límites estrictos pero con penalización por desbalance
    nombre_dimension = 'Tiempo'
    routing.AddDimension(
        transit_callback_index,
        0,  # Sin acumulación inicial artificial
        1000000,  # Capacidad máxima extendida para admitir rutas largas como Yucatán
        True,  # Iniciar acumulador en cero
        nombre_dimension
    )
    dimension_tiempo = routing.GetDimensionOrDie(nombre_dimension)
    dimension_tiempo.SetGlobalSpanCostCoefficient(100) # Minimiza la diferencia entre el viaje más largo y el más corto

    search_parameters = pywrapcp.DefaultRoutingSearchParameters()
    search_parameters.first_solution_strategy = (
        routing_enums_pb2.FirstSolutionStrategy.PATH_CHEAPEST_ARC
    )
    search_parameters.local_search_metaheuristic = (
        routing_enums_pb2.LocalSearchMetaheuristic.GUIDED_LOCAL_SEARCH
    )
    search_parameters.time_limit.seconds = 5

    solucion = routing.SolveWithParameters(search_parameters)

    rutas = {}
    if solucion:
        for vehiculo_id in range(num_vehiculos):
            index = routing.Start(vehiculo_id)
            ruta_vehiculo = []
            while not routing.IsEnd(index):
                ruta_vehiculo.append(manager.IndexToNode(index))
                index = solucion.Value(routing.NextVar(index))
            ruta_vehiculo.append(manager.IndexToNode(index))
            rutas[f"Vehículo {vehiculo_id + 1}"] = ruta_vehiculo
    return rutas

# Resolver el problema con 4 vehículos de forma inteligente
rutas_optimas = resolver_vrp_ultra_eficiente(num_vehiculos=4)

# Mostrar nuevos itinerarios balanceados y más eficientes
print("--- PLAN DE RUTAS INTELIGENTE Y ULTRA EFICIENTE ---\n")
colores_vrp = ["#E74C3C", "#2ECC71", "#3498DB", "#9B59B6"]
mapa_balanceado = folium.Map(location=[df['Latitud'].mean(), df['Longitud'].mean()], zoom_start=6)

# Añadir marcador del Cedis en León
folium.Marker(
    location=[df.iloc[0]['Latitud'], df.iloc[0]['Longitud']],
    popup="<b>Centro de Distribución (León)</b>",
    icon=folium.Icon(color="black", icon="briefcase")
).add_to(mapa_balanceado)

for i, (nombre_vehiculo, ruta) in enumerate(rutas_optimas.items()):
    color = colores_vrp[i % len(colores_vrp)]

    # Calcular métricas reales
    dist_tot = sum(matriz_distancia_rapida[ruta[k]][ruta[k+1]] for k in range(len(ruta)-1))
    tiempo_tot = sum(matriz_tiempo_rapida[ruta[k]][ruta[k+1]] for k in range(len(ruta)-1))

    ciudades_ruta = [df.iloc[idx]['Ciudad'] for idx in ruta]
    print(f"🚚 {nombre_vehiculo} ({len(ruta)-2} entregas):")
    print(f"   - Itinerario: {' -> '.join(ciudades_ruta)}")
    print(f"   - Métricas: {dist_tot:.1f} km | {tiempo_tot/60:.2f} horas de conducción")
    print("-" * 85)

    # Dibujar la ruta utilizando la geometría exacta de OSRM
    coords = [f"{df.iloc[idx]['Longitud']},{df.iloc[idx]['Latitud']}" for idx in ruta]
    url = f"https://router.project-osrm.org/route/v1/driving/{';'.join(coords)}?overview=full&geometries=geojson"
    try:
        resp = requests.get(url).json()
        if resp.get("code") == "Ok":
            geom = resp["routes"][0]["geometry"]["coordinates"]
            puntos_carretera = [[lat, lon] for lon, lat in geom]
            AntPath(
                locations=puntos_carretera,
                delay=1000,
                color=color,
                pulse_color="#FFFFFF",
                weight=5,
                opacity=0.85,
                tooltip=f"{nombre_vehiculo} ({tiempo_tot/60:.1f}h)"
            ).add_to(mapa_balanceado)
    except:
        pass

    # Agregar marcadores en el mapa
    for paso, idx in enumerate(ruta[1:-1], 1):
        fila = df.iloc[idx]
        folium.Marker(
            location=[fila['Latitud'], fila['Longitud']],
            popup=f"<b>{nombre_vehiculo}</b><br>{fila['Ciudad']}<br>Parada {paso}",
            icon=folium.Icon(color="blue", icon="info-sign")
        ).add_to(mapa_balanceado)

mapa_balanceado
```
